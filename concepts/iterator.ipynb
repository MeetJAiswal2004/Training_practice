## iterator  = an iterator is an object that can be iterated using next()

## iteration = it means accessing elements one by one from the collection (e.g iter())
## iterable = an iterable is an collection whose elements can be accessed one by one. (e.g. list and tuple, set, dict)

## iter() = it is used to create or obtain iterator
## next() = it retrives the next element

## when iterator has no more elements = error comes StopIteration

```python
## 1st example
lst = [1,2,3,4,5]

def iterator(lst):
    number = iter(lst)
    print(next(number))
    print(next(number))

print(iterator(lst))
```

```python
## 2nd example

lst1 = [1,2,3,4,5,6]

iterator = iter(lst1)

while True:
    try:
        num = next(iterator)
        print(num)
    except StopIteration:
        break
```

```python
## 3rd example

employees = ["meet","dev","raj"]

iterator = iter(employees)

print(next(iterator))
print(next(iterator))
print(next(iterator))
```

# __iter__()	 -  Returns the iterator itself and initializes its state
# __next__()	 -  Returns the next value
# StopIteration	 - Stops iteration when no values remain
