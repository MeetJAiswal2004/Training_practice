## decorator is used to modify or extend the behaviour of a function , without changing its actual code 

## 1st example 
def thirdnum(addition):
    def wrapper():
        result = addition()
        num3 = float(input("enter 3rd num:"))
        result = result + num3
        return result
    return wrapper


@thirdnum
def addition ():
    num1 = float(input("enter 1st num:"))
    num2 = float(input("enter 2nd num:"))
    result = num1 + num2
    return result


print(addition())

def decorator(func):
    def wrapper():
        print("before function")
        func()
        print("after function")
    return wrapper

@decorator
def greet():
    print("hello")

print(greet())
