- zip() is used to iterate over multiple list and pair the elements index wise

```python
a = [1,2,3,4]
b = ['meet','dev','raj','sagar']

result = list(zip(a,b))
print(result)
```

```python
for x, y in zip([1,2,3], ['a','b','c']):          # iterate multiple lists together
    print(x, y)
```

```python
x = [1,2,3,4]
y = [10,20,30,40]

for p,q in zip(x,y):
    print(p+q)
```
