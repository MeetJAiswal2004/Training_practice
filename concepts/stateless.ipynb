```python
## Stateless: The system processes each request independently. 


from fastapi import FastAPI

app = FastAPI()

@app.get("/add")
def add_numbers(a: int, b: int):
    return {"result": a + b}

## each requests acts idependently based on their parameters, 
# It does not need to rely on its stored session state for information from previous requests.

# GET /add?a=10&b=20  # 30
# GET /add?a=5&b=15   # 20
```
