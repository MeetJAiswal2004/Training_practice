```python
import time

def calculate_time(func):
    def wrapper(a,b):
        start = time.time()
        result = func(a,b)
        end = time.time()
        print(f"time:{end-start}")
        return result 
    return wrapper

@calculate_time
def add(a,b):
    result = a + b
    return result

add(10,20)
```

```python
def add_third_num(func):
    def wrapper():
       result = func()
       num3 = int(input("enter third num:"))
       result += num3
       return result
    return wrapper


@add_third_num
def add():
    num1 = int(input("enter first num:"))
    num2 = int(input("enter second num: "))
    result = num1 + num2
    return result

add()
```

```python
def total(*args):                  # collects extra positional args into a tuple
    return sum(args)

total(1, 2, 3, 4)    # -> 10

# **kwargs takes internally as dictionary
def show_info(**kwargs):           # collects extra keyword args into a dict
    for key, value in kwargs.items():
        print(key, value)

show_info(name="Alice", age=30)
```
