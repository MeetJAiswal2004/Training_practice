```python
import time

def calculate_time(func):
    def wrapper(a,b):
        start = time.time()
        result = func(a,b)
        end = time.time()
        print(f"time:{end-start}")
        return result 
    return wrapper

@calculate_time
def add(a,b):
    result = a + b
    return result

add(10,20)
```

```python
def add_third_num(func):
    def wrapper():
       result = func()
       num3 = int(input("enter third num:"))
       result += num3
       return result
    return wrapper


@add_third_num
def add():
    num1 = int(input("enter first num:"))
    num2 = int(input("enter second num: "))
    result = num1 + num2
    return result

add()
```
