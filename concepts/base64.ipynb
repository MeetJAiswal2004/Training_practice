#### Base64 is an encoding technique that converts binary data into a text format using 64 different characters.

##### Base64 contains 64 characters:
- A-Z → 26 uppercase letters
- a-z → 26 lowercase letters
- 0-9 → 10 digits
- '+' or '/' → two additional characters
'=' is used for padding

```python
import base64

text = "Hello"

print(text.encode())  ##output-> b'Hello' , here b indicates that the value is bytes rather than a normal Python string.

# Encode
encoded = base64.b64encode(text.encode()) ##Takes the bytes produced by text.encode() and Converts those bytes into Base64-encoded bytes.

print(encoded)     # output-> b'SGVsbG8='   

print(encoded.decode()) #output-> SGVsbG8=  , it Converts the Base64-encoded bytes into a normal Python string.

# Decode

#case 1
decoded = base64.b64decode(encoded) # base64.b64decode(encoded) converts the Base64-encoded bytes back into the original bytes.
print(decoded) # b'Hello'

#case 2
decoded = base64.b64decode(encoded).decode()  # Converts the original bytes back into a normal Python string.
print(decoded) # Hello , the original text is restored
```

encode() → Converts a string into bytes.

b64encode() → Converts bytes into Base64 format.

b64decode() → Converts Base64 data back into bytes.

### Use Cases of Base64


1. Sending Images Through APIs
- Sometimes, we need to send an image inside a JSON request body. Base64 can convert the image data into a string.

- Example:
{
    "image": "iVBORw0KGgoAAAANSUhEUg..."
}

-Use case: Sending image data from a frontend application to a FastAPI backend through JSON.


2. Email Attachments
 - Email systems use Base64 encoding to represent binary files, such as images and PDFs, in email messages.

 - Use case: Sending a PDF document as an email attachment.

3. Encoding Files for Data Transfer
- Base64 can represent binary files as text when a system only accepts text-based data.
- Use case: Transferring a small file through a text-based API.

4. JWT Authentication Tokens
- JSON Web Tokens (JWTs) use Base64URL encoding to represent their header and payload.

- Use case: Transmitting authentication information between a client and a backend.

- Important: JWT payloads are generally encoded, not encrypted. Do not store sensitive information in an ordinary JWT payload assuming it is secret.

5. Embedding Images in HTML or CSS
- Base64 can represent an image directly inside HTML or CSS instead of referencing a separate image file.

#### When Should You Use Base64?
- Use Base64: When binary data must be represented in a text-based format.
- Avoid Base64: When transferring large files unnecessarily, because it increases data size.
- Never use Base64 as security: It does not encrypt or protect sensitive data.
