- recursion is the phenomenon when function calling itself

```python
# example 1

lst = [1,2,3,4,5]
def total(lst):
    if len(lst) ==0:
        return 0
    
    return lst[0]+total(lst[1:])

print(total(lst))
```

```python
## factorial 
def factorial(n):
    if n == 0:   # because factorial 0 is one 
        return 1

    return n * factorial(n-1)

print(factorial(5))

```

```python
## fibonacci series is a sequence of numbers where every number is the sum of previous tweo numbers

def fibonacci_series(n):
    a, b = 0, 1
    series = []
    for _ in range(n):
        series.append(a)
        a, b = b, a + b
    return series

# Example: first 10 numbers
print(fibonacci_series(10))

n = 10
a, b = 0, 1

for _ in range(n):
    print(a, end=" ")
    a, b = b, a + b


def fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)

# Generate first 10 terms
n = 10
series = [fibonacci(i) for i in range(n)]
print(series)
```

```python
# flatten deep nested list

lst = [1,2,[3,4,[5,6,[7]]]]
flat =[]
def flatten(lst,flat):
    
    for n in lst:
        if type(n) == list:
            flatten(n,flat) # recursion
        else:
            flat.append(n)
    return flat

print(flatten(lst,flat))
```

```python
lst =[1, 2, 3, 4, 5, 6, 7]

def nested(lst):
    if len(lst)==0:
        return lst
    
    return [lst[:2]+nested(lst[2:])]
print(nested(lst))  
```
