## decorator is used to modify or extend the behaviour of a function , without changing its actual code 

## Important Decorator Rules

| Concept | Remember |
|---|---|
| Decorator application | Bottom to top |
| Wrapper execution | Outer to inner |
| Wrapper completion | Inner to outer |
| `func(*args, **kwargs)` | Passes arguments to the original function |
| `return func(...)` | Preserves the returned result |
| `functools.wraps(func)` | Preserves function metadata |
| Decorator factory | Accepts configuration arguments |
| Closure | Wrapper remembers the original function |

- Question: Why do we define a nested wrapper function inside a decorator (like deco1 or deco2) instead of writing the logic directly inside the decorator function?

- answer :Because @deco1 and @deco2 runs immediately when the function(show) is defined, not when it is called 
             
             (show = deco1(deco2(show))):

- - Delayed & repeated execution: If the logic were written directly inside deco1, it would execute once right at definition time and never again. The nested wrapper delays execution so the logic runs only when—and every time—the decorated function is called.

- - Returning a callable: Python replaces the original function with whatever deco1 returns. Without returning wrapper, deco1 returns None, causing a TypeError when you later try to call the function.

- - Closure access: Being nested, wrapper remembers the original func passed to deco1 and can run code before and after calling func().

##### example 1

- Decorator chaining and execution order

                           show =deco1(deco2(show)) (bottom to top)

- always execute first decorator(i.e deco1) in the below case ,bcz python is interpreted language

```python
def deco1(func):
    def wrapper():
        print("D1 start")
        func()
        print("D1 end")
    return wrapper

def deco2(func):
    def wrapper():
        print("D2 start")
        func()
        print("D2 end")
    return wrapper

@deco1
@deco2
def show():
    print("Hello") # here hello is print in the middle of the last decorator

show()
```

- always execute first decorator(i.e deco2) in the below case ,bcz python is interpreted language

```python
@deco2
@deco1
def show():
    print("Hello")

show()
```

- decorator works on bottton up approach , for the below question order is:
  
         number = add_10(multiply_2(number))

```python
def add_10(func):
    def wrapper():
        return func() + 10  #  3rd 10+10=20 (output)
    return wrapper

def multiply_2(func):
    def wrapper():
        return func() * 2   # 2st 5*2=10
    return wrapper

@add_10
@multiply_2
def number():
    return 5                # 1st 5 (actual value)

print(number())
```

- Decorator with *args and **kwargs

- Decorator factories

 - @wraps from functools is used in decorators to preserve the original function's metadata.

```python
## with wraps
from functools import wraps
def repeat(n):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
@repeat(3)
def greet():
    print("Hello")

greet()
```

```python
## without wraps
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
@repeat(3)
def greet():
    print("Hello")

greet()
```

- Prefix decorator

```python
## with wraps
from functools import wraps
def prefix(text):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            return f"{text}: {func(*args, **kwargs)}"
        return wrapper
    return decorator


@prefix("INFO")
def message():
    return "Server started"

print(message())
```

```python
## without wraps
def prefix(text):
    def decorator(func):
        def wrapper(*args,**kwargs):
            return f"{text} {func(*args,**kwargs)}"
        return wrapper
    return decorator

@prefix("INFO")
def message():
    return "Server started"

print(message())
```

- Role-based decorator

```python
from functools import wraps
current_role = "admin"

def check_role(role):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            if current_role != role:
                raise PermissionError("Access denied")
            return func(*args, **kwargs)
        return wrapper
    return decorator

@check_role("admin")
def delete_user():
    print("User deleted")

delete_user()
```

- delay decorator

```python
import time
def delay(seconds):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            time.sleep(seconds)
            start = time.time()
            result =func(*args, **kwargs)
            end = time.time()
            print("total time:", end-start)
            return result
        return wrapper
    return decorator

@delay(2)
def task():
    print("Task completed")

task()
```

- Retry decorator

```python
def retry(n):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            for attempt in range(n):
                try:
                    return func(*args, **kwargs)
                except Exception:
                    if attempt == n - 1:
                        raise
        return wrapper
    return decorator

@retry(3)
def connect():
    print("Trying connection")
    raise ValueError("Connection failed")
    
```

- Preserve function metadata

```python
# preserving metadata using functools.wraps
def decorator(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper

@decorator
def calculate():
    """Calculate something."""
    return 100

calculate()
```

- Decorator with exception handling

```python
def handle_exception(func):
    def wrapper(*args,**kwargs):
        try:
            result= func(*args,**kwargs)
            return result
        except ZeroDivisionError as e:
            print(f"cannot divide by 0: {e}")
        finally:
            print("execution complete")
    return wrapper


@handle_exception
def divide(a, b):
    return a / b

divide(10, 0)
```

```python
## second method
from functools import wraps
def handle_exception(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        try:
            return func(*args, **kwargs)
        except Exception as e:
            print(f"Error occurred: {e}")
    return wrapper

@handle_exception
def divide(a, b):
    return a / b

divide(10, 0)
```

- Execution time decorator

```python
def timer(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"Execution time: {end - start:.2f} seconds")
        return result
    return wrapper

@timer
def task():
    total = sum(range(1000000))
    return total

task()
```

- Decorator order with arguments

```python
def deco1(func):
    @wraps(func)
    def wrapper(x):
        print("D1 start")
        result = func(x) + 10
        print("D1 end")
        return result
    return wrapper


def deco2(func):
    @wraps(func)
    def wrapper(x):
        print("D2 start")
        result = func(x) * 2
        print("D2 end")
        return result
    return wrapper


@deco1
@deco2
def number(x):
    return x


print(number(5))
```

- Cache decorator

```python
def cache_result(func):
    cache = {}

    @wraps(func)
    def wrapper(*args, **kwargs):
        key = (args, tuple(sorted(kwargs.items())))

        if key not in cache:
            cache[key] = func(*args, **kwargs)

        return cache[key]
    return wrapper


@cache_result
def square(n):
    print("Calculating...")
    return n * n

print(square(5))
print(square(5))
```

- Validate positive numbers

```python
def positive_only(func):
    @wraps(func)
    def wrapper(n):
        if n <= 0:
            raise ValueError("Number must be positive")
        return func(n)
    return wrapper

@positive_only
def square(n):
    return n * n

print(square(5))
print(square(-5))
```

- Count function calls

```python
def count_calls(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        wrapper.calls += 1
        return func(*args, **kwargs)

    wrapper.calls = 0
    return wrapper

@count_calls
def greet():
    print("Hello")

greet()
greet()
greet()

print(greet.calls)
```

- Build a complete decorator chain

```python
def uppercase(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs).upper()
    return wrapper


def add_prefix(text):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            return f"{text}: " + func(*args, **kwargs)
        return wrapper
    return decorator


def reverse_text(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)[::-1]
    return wrapper


@uppercase
@add_prefix("INFO")
@reverse_text
def message():
    return "python"

print(message())
```
