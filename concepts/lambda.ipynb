- python identifies this function internally as lambda ,this is why it is called as an anounomus function, 

- we use lambda function for performing short operations 
- we do not need to provide function name over here
- lambda function can only perform one expression at a time 
- return keyword is not required

- formate of map(),filter(),reduce()

e.g. map(function,iterable) 

```python
## lambda function for square,adding,multiplying of a particular number

square = lambda x : x ** 2
print(square(5))

add = lambda x,y : x +y
print(add(2,3))

multiply= lambda x,y : x*y
print(multiply(2,3))

```

```python
## lambda function for two's table
num = int(input("enter a number"))
table = list(map(lambda x : x*2,range(1,11)))
print(table)
```

```python
## conditional statement in lambda function
x = int(input("enter a number:"))
result = (lambda x : "positive" if x > 0 else "negative" if x < 0 else "zero") (x)

print(result)
```

```python
nums = [2, 5, 7, 8, 10, 12]

sqaured_num = list(map((lambda x : x ** 2),nums))

print(sqaured_num)
```

```python
## filter those numbers whose square is > 50
nums = [2, 5, 7, 8, 10, 12]

result = list(filter(lambda x: x > 50,
                     map(lambda x: x ** 2, nums)))

print(result)
# [64, 100, 144]
```

```python
## find the second largest number without sort()

nums = [2, 5, 7, 8, 10, 12]

largest =max(nums) #12
second_largest = max(filter(lambda x :x < largest, nums))

print(second_largest)
```

```python
## sort the list of tuple based on their second element 
data = [(1, 50), (2, 20), (3, 40), (4, 10)]
# Expected: [(4, 10), (2, 20), (3, 40), (1, 50)]

data = [(1, 50), (2, 20), (3, 40), (4, 10)]

result = sorted(data, key=lambda x: x[1])

print(result)
# [(4, 10), (2, 20), (3, 40), (1, 50)]

```

```python
## Convert a list of strings into uppercase
names = ["meet", "amit", "rahul", "priyanka", "john"]
# Expected: ['RAHUL', 'PRIYANKA']

upper = list(map(lambda x : x.upper(),names))
print(upper)
```

```python
# Convert a list of strings into uppercase and filter strings having length greater than 4

names = ["meet", "amit", "rahul", "priyanka", "john"]


upper = list(filter((lambda x: len(x)> 4),map(lambda x : x.upper(),names)))
print(upper)
```

```python
## find the product of all elemnts
from functools import reduce

nums = [2, 3, 4, 5]

result = reduce(lambda x, y: x * y, nums)

print(result)
# 120
```

```python
## find the numbers which are divisible by 3 and 5
nums = [15, 20, 30, 45, 50, 60, 75]

result = list(filter((lambda x : x%3 == 0 and x%5 == 0),nums))

print(result)
```

```python
#Sort words by length and then alphabetically
words = ["banana", "cat", "apple", "dog", "kiwi"]

result = sorted(words, key=lambda x: (len(x), x))

print(result)
# ['cat', 'dog', 'kiwi', 'apple', 'banana']
```

```python
data = [[10, 20, 30], [5, 50, 15], [100, 40, 60]]

result = list(map(lambda x: max(x), data))

print(result)
# [30, 50, 100]
```
