#### formate of map(),filter(),reduce()

e.g. map(function,iterable) ,similarly for filter() and reduce()

(note)- above three functions must takes two arguments

map() → transforms every element of an iterable.

filter() → selects specific elements based on some condition.

reduce() → it is used when we want to compare /combiine multiple elements into a single final result.

- lambda function for square,adding,multiplying of a particular number

```python
square = lambda x : x ** 2
print(square(5))

add = lambda x,y : x +y
print(add(2,3))

multiply= lambda x,y : x*y
print(multiply(2,3))

```

- filter those numbers whose square is > 50

```python
nums = [2, 5, 7, 8, 10, 12]

result = list(filter(lambda x: x > 50,
                     map(lambda x: x ** 2, nums)))

print(result)
# [64, 100, 144]
```

- find the second largest number without sort()

```python
nums = [2, 5, 7, 8, 10, 12]

largest =max(nums) #12
second_largest = max(filter(lambda x :x < largest, nums))

print(second_largest)
```

- sort the list of tuple based on their second element 

```python
data = [(1, 50), (2, 20), (3, 40), (4, 10)]
# Expected: [(4, 10), (2, 20), (3, 40), (1, 50)]

data = [(1, 50), (2, 20), (3, 40), (4, 10)]

result = sorted(data, key=lambda x: x[1])

print(result)
# [(4, 10), (2, 20), (3, 40), (1, 50)]

```

- Convert a list of strings into uppercase

```python

names = ["meet", "amit", "rahul", "priyanka", "john"]
# Expected: ['RAHUL', 'PRIYANKA']

upper = list(map(lambda x : x.upper(),names))
print(upper)
```

- Convert a list of strings into uppercase and filter strings having length greater than 4

```python
names = ["meet", "amit", "rahul", "priyanka", "john"]

upper = list(filter((lambda x: len(x)> 4),map(lambda x : x.upper(),names)))
print(upper)
```

- Sum of positive numbers using filter() and reduce()

```python
from functools import reduce
nums = [10, -5, 20, -10, 30]

result = reduce(lambda x,y : x+y ,list(filter(lambda x: x>0 ,nums)))
print(result)
```

```python
## second method

nums = [-10, -5, -20, -10, -30]

positive = filter(lambda x: x > 0, nums)

result = reduce(lambda x, y: x + y, positive,0) ## here 0 is used when there is no +ve num in the list (prevent from value error)

print(result)
# 60
```

- Sort employees by salary descending

```python
employees = [
    {"name": "Meet", "salary": 30000},
    {"name": "Amit", "salary": 50000},
    {"name": "Rahul", "salary": 40000}
]

result = sorted(employees, key = lambda x : x['salary'], reverse=True)

print(result)
```

- Find duplicate elements using filter()

```python
nums = [1, 2, 3, 2, 4, 5, 1, 6]

# set() removes repeated results. This is a learning solution; count() makes it O(n²) overall.
result = list(set(filter(lambda x: nums.count(x) > 1, nums)))

print(sorted(result))
```

- Flatten a nested list using reduce()

```python
data = [[1, 2], [3, 4], [5, 6]]

result = reduce(lambda x, y: x + y, data)

print(result)

```

```python
## second method
data = [[1, 2], [3, 4], [5, 6],[]]

result = reduce(lambda x, y: x + y, data, []) ## [] handles the empty list in the nested list

print(result)
```

- Find employee with highest salary using reduce()

```python
employees = [
    {"name": "Meet", "salary": 30000},
    {"name": "Amit", "salary": 50000},
    {"name": "Rahul", "salary": 40000}
]

result = reduce(
    lambda x, y: x if x["salary"] > y["salary"] else y,
    employees
)

print(result)
```

- Sort dictionaries by age ascending and salary descending

```python
employees = [
    {"name": "Meet", "age": 23, "salary": 30000},
    {"name": "Amit", "age": 25, "salary": 50000},
    {"name": "Rahul", "age": 23, "salary": 40000},
    {"name": "Priya", "age": 25, "salary": 60000}
]

result = sorted(employees, key=lambda x: (x["age"], -x["salary"]))

print(result)
```

- Find the sum of digits of each number

- Find the longest word using reduce()

```python
words = ["python", "java", "javascript", "sql"]

result = reduce(
    lambda x, y: x if len(x) > len(y) else y,
    words
)

print(result)
```

- Group numbers into even and odd

```python
nums = [1, 2, 3, 4, 5, 6, 7, 8]

result = {
    "even" : list(filter(lambda x:x%2==0,nums)),
    "odd" : list(filter(lambda x : x%2 != 0 ,nums))
}

print(result)
```

- Even numbers ascending, odd numbers descending

```python
nums = [5, 2, 8, 1, 9, 4, 7, 6]

result = sorted(
    nums,
    key=lambda x: (x % 2, -x if x % 2 else x)
)

print(result)
# [2, 4, 6, 8, 9, 7, 5, 1]
```

- Find the most frequent element using reduce()

```python
nums = [1, 2, 2, 3, 3, 3, 4, 4]

result = reduce(
    lambda x, y: x if nums.count(x) >= nums.count(y) else y,
    nums
)

print(result)
```

- Find pairs whose sum equals the target

```python
from itertools import combinations

nums = [2, 4, 3, 5, 7, 8, 1]
target = 9

result = list(filter(
    lambda x: sum(x) == target,
    combinations(nums, 2)
))

print(result)
```

- Find students whose average marks are greater than 75

```python
students = [
    {"name": "Meet", "marks": [80, 90, 85]},
    {"name": "Amit", "marks": [60, 70, 65]},
    {"name": "Rahul", "marks": [90, 95, 85]}
]

result = list(map(
    lambda x: x["name"],
    filter(
        lambda x: sum(x["marks"]) / len(x["marks"]) > 75,
        students
    )
))

print(result)
# ['Meet', 'Rahul']
```
