- Generator produces values one at a time, rather than storing all values in memory.they are memory efficient and generate values lazily ,one at a time

- Generators are useful for lazy, incremental data processing

- yield is used to create a generator function. It allows the function to return a value and pause its execution, saving its state for the next call. When the generator is called again, it resumes execution from where it left off.

# return	                                      yield
# Returns a value and exits the function     ||    Produces a value and pauses the function
# Function execution ends	                 ||    Function execution can resume
# Usually returns a normal value	         ||    Makes the function a generator
# Doesn't preserve execution for another call||	   Preserves local state between yields

# Real-world use cases for generators include:

# Application	         How generators help
# Large CSV files	     Process records one by one
# Machine learning	     Load batches of training data
# NLP	                 Process large text datasets
# APIs	                 Stream data instead of returning everything at once
# Log processing	     Read and process logs incrementally
# Data pipelines	     Pass processed records between stages

```python
# example 1 
def numbers():
    for i in range(5):
        yield i 

for n in numbers():
    print(n)
```

```python
## 2nd example
def numbers():
    yield 10
    yield 20
    yield 30

result = numbers()

print(next(result))
print(next(result))
print(next(result))
```

```python
## 3rd example

# print(next(gen))   yield pauses the generator.
# When next() is called again, execution resumes from where it paused.
# After the generator finishes, Python raises StopIteration.

def my_generator():
    print("First")
    yield 1

    print("Second")
    yield 2

    print("Third")
    yield 3


gen = my_generator()


print(next(gen))
print(next(gen))
```
