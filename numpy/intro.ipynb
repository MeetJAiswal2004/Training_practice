##### numpy provides fast numerical operations and also handles array/metrices for scientific computing



import numpy as np

```python
import numpy as np

arr=np.array([10,20,30,40,50])
print(type(arr))
```

```python
arr = np.array([[1, 2, 3],
                 [4, 5, 6]])

print(arr.sum(axis=0))   # -> [1+4, 2+5, 3+6] = [5,7,9]   (collapse rows, sum down each column)
print(arr.sum(axis=1))   # -> [1+2+3, 4+5+6] = [6,15]     (collapse columns, sum across each row)
```

- Broadcasting is the most important NumPy concept

It's how NumPy performs arithmetic on arrays of different shapes without explicit loops.

```python
a = np.array([1, 2, 3])          # shape (3,)
b = np.array([[1], [2], [3]])    # shape (3,1)

a + b
#shape (3,) broadcasts with (3,1) -> result shape (3,3)
```

### Random number generation

```python
print(np.random.seed(42))              # reproducibility (legacy API)

print(np.random.rand(3, 2))       # uniform [0,1), shape (3,2)
print(np.random.randn(3, 2))          # standard normal distribution
print(np.random.randint(0, 10, size=5)) # random ints in [0,10)
print(np.random.choice([1,2,3,4], size=3, replace=False) ) # random sample
```

```python
a = np.array([1, 2, np.nan, 4])

np.isnan(a)          # array([False, False, True, False])
np.nanmean(a)        # mean ignoring NaN
np.nansum(a)         # sum ignoring NaN
np.nanmax(a)         # max ignoring NaN
a[~np.isnan(a)]      # filter out NaNs
```

```python
a = np.array([1, 2, 2, 3, 4])

print(np.unique(a))                          # [1, 2, 3, 4]
print(np.unique(a, return_counts=True) )     # values + counts

print(np.intersect1d([1,2,3], [2,3,4]))      # [2, 3]
print(np.union1d([1,2,3], [2,3,4]))          # [1,2,3,4]
print(np.setdiff1d([1,2,3], [2,3]))          # [1]
print(np.isin([1,2,3,4], [2,4]))             # [False, True, False, True]
```
