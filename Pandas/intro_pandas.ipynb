- Pandas is one of the most important Python libraries for data analysis, data manipulation, data cleaning, and data processing.

(or)

- Pandas is used for data manipulation, data cleaning, data tranformation, and analysizing using dataframe and series

#### Important terms

| Terminology | Meaning |
|---|---|
| Data | Raw information |
| Dataset | Collection of data |
| DataFrame | 2D tabular data structure |
| Series | 1D labeled data structure |
| Index | Labels used to identify rows |
| Column | Vertical collection of data |
| Row | Horizontal collection of data |
| Axis | Direction of operation |
| Missing values | Values that are absent |
| Data types | Types of stored values |

#### imp methods

| Method | Purpose |
|---|---|
| `loc[]` | Label-based selection,, Select rows and columns using their labels.|
| `iloc[]` | Position-based selection,, Select rows and columns using their integer positions. |
| `at[]` | Access a single value by label |
| `iat[]` | Access a single value by position |
| `query()` | Filter rows using expressions |
| `filter()` | Select rows or columns by labels |
| `set_index()` | Set a column as the index |
| `reset_index()` | Reset the index |

 ### Important difference:
- loc works with labels.
- iloc works with integer positions.

# imp methods

- .map() → Series only, element-wise, good for value substitution via dict.
- .apply() → Series (element-wise) OR DataFrame (row/column-wise via axis).
- .applymap() → DataFrame, element-wise on every cell (renamed to .map() on DataFrames in pandas 2.1+).
