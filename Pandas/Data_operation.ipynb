## Pandas Data operation and setup through kaggle


### first method for loading dataset from kaggle

- #### How it works: 
Directly reads the CSV file stored in your local directory into memory as a DataFrame.


- #### Key Advantages:
- Fastest & Offline: Reads directly from local storage with zero network latency; does not require internet connection.
- Simplicity: No external API keys or extra dependencies needed.


- #### Limitations:
- Not portable: If you share the notebook with someone else, it will fail unless they also manually download and place the CSV in the exact same folder.
- No automatic updates: If the dataset is updated upstream on Kaggle, your local copy will remain outdated.

### second method for loading dataset from kaggle


- ### How it works: 
Queries Kaggle’s API to download the dataset into your system's cache directory, then points read_csv to the cached path.

- ### Key Advantages:
- Portability & Reproducibility: Anyone who runs your notebook anywhere will automatically download the dataset without needing manual file sharing.
- Smart Caching: It only downloads over the internet once; subsequent runs instantly read from your local cache.
- Version control: Always pulls the latest version (or specific tagged version) directly from the author.


- ### Limitations:
- Requires Internet: Needs internet connectivity on the first execution.
- Extra Dependency: Requires kagglehub / Kaggle API setup.

```python
## import preferred module 
import kagglehub

## set path variable for the dataset 
path = kagglehub.dataset_download('shantanugarg274/sales-dataset')
print(path)

## import os for verifying the dataset path
import os
print(os.listdir(path))

## join the kaggle's dataset path with the os to perform operations
import pandas as pd
df=pd.read_csv(os.path.join(path, 'Sales Dataset.csv'))
df.head(5)
```

### Cleaning and Handling Missing Data

```python
# s-1 download dataset from kaggle using url
!kaggle datasets download -d sohaibdevv/ai-and-data-science-job-market-2025-2026

# s-2 check if it is installed in the file (temporary)
!ls

# s-3 unzip dataset
!unzip ai-and-data-science-job-market-2025-2026.zip

# s-4 using dataframes with pandas
import pandas as pd
df = pd.read_csv('AI_Job_Market_Trends_2026.csv')
df.head(2)
```

```python
# Check for missing values per column
print("Missing values per column:")
display(df.isnull().sum())
```

```python
# As there are no missing values in this dataset, let's create a copy and introduce some for demonstration.
df_temp = df.copy()

# make manipulation in the copy of dataset (best practice for manipulations on copy of datasets)
df_temp.loc[0, 'salary'] = None
df_temp.loc[1, 'company_size'] = None

print("\nMissing values in temporary DataFrame:")
display(df_temp.isnull().sum())

# # Drop rows with any missing values using dropna( ) function especially focuses on missing values
df_dropped_na = df_temp.dropna()
print(f"\nShape after dropping NA: {df_dropped_na.shape} (original: {df.shape})")

#to drop specific column's null value
df_dropped_na = df_temp.dropna(subset=['salary'], inplace=True)


# Fill missing values (e.g., with a specific value or the mean/median)
df_filled_na = df_temp.fillna({'salary': df_temp['salary'].mean(), 'company_size': 'Unknown'})
print("\nMissing values after filling NA:")
display(df_filled_na.isnull().sum())
```

#### checking for duplicate rows 

```python
# Check for duplicate rows
print(f"Number of duplicate rows: {df.duplicated().sum()}")

# Drop duplicate rows (if any)
df_no_duplicates = df.drop_duplicates()
print(f"Shape after dropping duplicates: {df_no_duplicates.shape}")
```
