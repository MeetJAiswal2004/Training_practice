### DataFrame (2D labeled table)

DataFrame is a 2d labelled tabular data structure

##### dataframe only methods 

| Method                            | Purpose                                                     | Example                                         |
| --------------------------------- | ----------------------------------------------------------- | ----------------------------------------------- |
| **`.info()`**                     | Summary of column types, non-null counts, and memory usage. | `df.info()`                                     |
| **`.query()`**                    | Filters rows using an expression string.                    | `df.query("age > 25 and salary > 30000")`       |
| **`.select_dtypes()`**            | Subsets columns by data type (e.g., `include='number'`).    | `df.select_dtypes(include='int64')`             |
| **`.melt()`**                     | Unpivots a DataFrame from wemp_ide format to long format.       | `df.melt(emp_id_vars=['name'])`                     |
| **`.pivot()` / `.pivot_table()`** | Reshapes data into a cross-tabulated pivot table.           | `df.pivot_table(values='salary', index='city')` |
| **`.merge()` / `.join()`**        | Database-style SQL joins between two DataFrames.            | `df1.merge(df2, on='emp_emp_id')`                   |
| **`.insert()`**                   | Inserts a new column at a specific index position.          | `df.insert(0, 'emp_id', [1, 2, 3])`                 |
| **`.assign()`**                   | Creates or overwrites columns returns a new DataFrame.      | `df.assign(bonus=df['salary'] * 0.1)`           |

```python
import pandas as pd
df = pd.DataFrame({
    "name":["meet","dev","raj"],
    "age": [22,24,20],
    "city":["bu","ny","hk"] 
  }
)
print(df)
display(df)
```

```python
from IPython.core.display_functions import display
import numpy as np


print("From dict of lists")
a = pd.DataFrame({'a': [1,2], 'b': [3,4]})
display(a)

print("From list of dicts")
b = pd.DataFrame([{'a':1,'b':2}, {'a':3,'b':4}])
display(b)

print("From list of lists (with explicit columns)")
c = pd.DataFrame([[1,2],[3,4]], columns=['a','b'])
display(c)

print("From NumPy array")
d = pd.DataFrame(np.random.rand(3,2), columns=['x','y'])
display(d)

print("From Series")
e = pd.DataFrame(pd.Series([1,2,3]), columns=['val'])
display(e)

print("Empty DataFrame with defined columns")
f = pd.DataFrame(columns=['a', 'b', 'c'])
display(f)
```

| Parameter     | Purpose                                 |
|---------------|-----------------------------------------|
| `sep`         |  Defines the separator                  |
| `header`      | Defines which row contains column names |
| `index_col`   | Sets a column as DataFrame index        |
| `usecols`     | Selects specific columns                |
| `dtype`       | Defines column data types               |
| `na_values`   | emp_identifies custom missing values        |
| `parse_dates` | Converts columns into datetime          |
| `nrows`       | Reads a specific number of rows         |
| `encoding`    | Defines character encoding              |
| `to_csv()`    | Exports DataFrame into CSV              |
| `index=False` | Prevents writing the DataFrame index    |

```python
import csv

data = [
    ["emp_id", "a", "b", "date_col", "city"],
    [1, "Meet", 5000, "2026-01-10", "Bhopal"],
    [2, "Amit", 6000, "2026-01-11", "Indore"],
    [3, "NA", 7000, "2026-01-12", "Delhi"],
    [4, "Ravi", "?", "2026-01-13", "Mumbai"],
    [5, "Priya", 8000, "2026-01-14", "Pune"],
    [6, "John", 9000, "2026-01-15", "Chennai"]
]

with open("file.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file, delimiter=";")
    writer.writerows(data)

print("CSV file created successfully!")
```

```python
df = pd.read_csv("file.csv")
df = pd.read_csv("file.csv", sep=';', header=0, index_col=0,
                  usecols=["emp_id", "a", "b", "date_col"], dtype={'a': str}, na_values=['NA', '?'],
                  parse_dates=['date_col'], nrows=1000, encoding='utf-8')

df.to_csv("out.csv", index=False)

display(df)
```

```python
import pandas as pd

df1 = pd.DataFrame({
    'emp_emp_id':[1,2,3],
    'name':['meet','dev','akki'],
    'age':[22,24,20]
})
display(df1)

df2=pd.DataFrame({
    'emp_emp_id':[1,2,3],
    'salary':[500000,450000,20000],
    'department':['ai','commerce','bio']
})
display(df2)

display(df1.merge(df2))
```

```python
df_mer=pd.merge(df1, df2, on='emp_id', how='inner')    # inner join (default)
display(df_mer)

df_merg=pd.merge(df1, df2, on='emp_id', how='left')     # left join
display(df_merg)

df_merge=pd.merge(df1, df2, on='emp_id', how='right')    # right join
display(df_merge)

df_merger=pd.merge(df1, df2, on='emp_id', how='outer')    # full outer join
display(df_merger)


# df_m = pd.merge(df1, df2, left_on='emp_id1', right_on='emp_id2')   # different key column names
# display(df_m)

# df_me = pd.merge(df1, df2, on='emp_id', suffixes=('_left', '_right'))  #
# display(df_me)
```

```python

df = pd.DataFrame({
    'name':['meet','dev','akki'],
    'age':[22,24,20]
})

df.insert(0, 'emp_id', [1, 2, 3]) 
display(df)

df_k=df.assign(bonus=df['age'] + 2) ## Creates or overwrites columns returns a new DataFrame.
display(df_k)

df_g = df.drop(0) ## drop specific row at whole index
display(df_g)

```

```python
df_concat = pd.concat([df1, df2]) 
display(df_concat)                   # stack rows (axis=0 default)
df_con = pd.concat([df1, df2], axis=1) 
display(df_con)
```

axis parameter refers to the row/column operation 
- axis = 0 , means column-wise operation (top-down) vertically
- axis = 1 , means row-wise operation(left to righ) horizontally

```python
df_age = df['age'].apply(lambda x: x * 2) # element-wise on a Series
display(df_age)             

df_row = df.apply(lambda row: row['age'] + 1, axis=1)      # row-wise across DataFrame
display(df_row) 

df_max = df.apply(lambda col: col.max(), axis=0)    # column-wise (default axis=0)
display(df_max) ## returns max value across each column
```

```python
import pandas as pd

df = pd.DataFrame({
    'name':['meet','dev','yash','rohan'],
    'age':[22,24,23,40],
    'city':['burhanpur','burhanpur','ujjain','nepal']
})

df_groupby=df.groupby('city')['age'].mean()
display(df_groupby)

df_group_agg=df.groupby('city').agg({'age': 'mean', 'name': 'count'})
display(df_group_agg)

df_result=df.groupby('city').agg(
    avg_age=('age', 'mean'),
    total=('name', 'count')
)
display(df_result)   
```
