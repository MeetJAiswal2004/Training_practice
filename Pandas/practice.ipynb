```python
import kagglehub

path = kagglehub.dataset_download('shantanugarg274/sales-dataset')
print(path)

import os
print(os.listdir(path))

file_path= os.path.join(path, 'Sales Dataset.csv')

import pandas as pd 
df = pd.read_csv(file_path)
display(df.head(3))
```

```python
display(df['Amount']) ## display only specific value

display(df[['Profit','City']]) ## detecting two columns and also access more columns similary at once
```

```python
display(df.loc[0:1,'Category']) ## label based selection ,also accessing rows via index range values
display(df.loc[1]) # accessing only single row using index value
display(df.loc[1,'Profit']) # accessing only single row of a particular column
display(df.loc[1,['Category','Profit']]) # accessing specific row of two or more columns 
```

```python
display(df.iloc[0:1, 4]) ## position based selection ,also accessing columns via integer value
display(df.iloc[1,2]) # accessing only single row of a particular column via integer
display(df.iloc[1,[4,2]]) # accessing specific row of two or more columns 
```

```python
display(df.at[1,'Profit'])
display(df.iat[1,3])  ## using 'iat' Access one specific cell using integer row and column positions.
```
