### Series (1D labeled array)

Series = 1d labelled data structure


##### below are the series only methods

| Method / Accessor | Purpose | Example |
|---|---|---|
| `.unique()` | Returns unique values as a NumPy array. (DataFrame does not have `.unique()`) | `s.unique()` |
| `.value_counts()` | Returns counts of unique values in a Series | `s.value_counts()` |
| `.to_frame()` | Converts a 1D Series into a 2D DataFrame | `s.to_frame(name='col_name')` |
| `.str.*` | Performs vectorized string operations (e.g., `.upper()`, `.contains()`) | `s.str.upper()` |
| `.dt.*` | Accesses datetime properties (e.g., `.dt.year`, `.dt.day_name()`) | `s.dt.month` |
| `.cat.*` | Performs categorical operations and accesses category properties | `s.cat.categories` |
| `.is_unique` | Checks whether all values in a Series are unique (`bool`) | `s.is_unique` |
| `.is_monotonic_increasing` | Checks whether values are in non-decreasing order | `s.is_monotonic_increasing` |
| `.argmax()` / `.argmin()` | Returns the integer position of the maximum / minimum value | `s.argmax()` |

- NOTE

Accessors like .str and .dt operate on individual columns (Series). To apply them to a DataFrame column, you must select the column first: 
`df['email'].str.lower()`.

```python
import pandas as pd

data = [10,80,20,40]

s = pd.Series(data)

print(s)
display(s)


s.unique() ## returns only unique values

s.value_counts() ## frequency of each element

s.to_frame(name='col_name') ## convert 1d series into 2d dataframe

s.is_unique ## True/Fase, if series containing all unique values

s.is_monotonic_increasing ## True/False, if the series elements are in continuous increasing nature

s.argmax() ## returns the index value of max element

s.argmin() ## returns the index value of min element


```

```python
s = pd.Series(['meet', 'dev', 'raj'], index=['a', 'b', 'c'], name='values')
print(s)

s.str.upper()
```
