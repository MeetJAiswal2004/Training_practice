##### Rag or Retrieval-Augmented-Generation is an AI framwork that imporves Large language model accuracy and efficiency by retrieving extrnal data to it, so that LLM not only answer with the static data which is gained by the LLm during its pretraining but also answer with the updated and dynamic data using RAG pipeline.
#####            so in this way RAG system acts as a external knowledge Base for our llm.

- RAG is mainly used to reduce hallucination from an AI model by using RAG's similarity search functionality.

- RAG is the process of optimizing the output of LLM

There are various types of RAG systems.
- `Normal RAG`: It is limited to text-based documents,it means it only supports text-based embeddings.

- `Multimodal RAG`: It is not limited to text-based documents. It supports embeddings based on text, images, and audio.

- `Naive RAG`: A basic RAG system that follows the process: Query → Retrieve documents → Generate an answer using an LLM.

- `Advanced RAG`: Uses techniques such as query rewriting, improved chunking, reranking, and metadata filtering to enhance retrieval accuracy.

- `Modular RAG`: Allows individual components, such as retrieval, reranking, and generation, to be independently modified or replaced.

- `Graph RAG`: Uses a knowledge graph to retrieve information based on relationships between entities.

- `Hybrid RAG`: Combines vector search and keyword-based search (BM25) to retrieve relevant information.

- `Agentic RAG`: An AI agent decides what information to retrieve, which tools or queries to use, and can perform multiple retrieval steps to generate a better answer.

## Working

##### RAG follows a structured workflow where a query is processed, relevant information is retrieved and a final response is generated using both retrieved data and model knowledge.

- `Query Processing`: The input query is first pre-processed and prepared for further steps, ensuring it is in a suitable form for embedding.

- `Embedding Model`: The query is passed through an embedding model that converts it into a vector capturing its semantic meaning.

- `Vector Database Retrieval`: This vector is used to search a vector database to find documents that are most similar to the query.

- `Retrieved Contexts`: The system retrieves the documents that are closest to the query. These documents are then forwarded to the generative model to help it craft a response.

- `LLM Response Generation`: The LLM combines the original query with the retrieved context to generate a coherent and accurate response.

- `reponse`: The final response integrates both the model’s internal knowledge and the retrieved information, making it more relevant and up-to-date.
