/*
========================================================
                SQL CLAUSES CHEAT SHEET
========================================================

1. SELECT
   - Used to retrieve data from a table.

2. DISTINCT
   - Used to return only unique values and remove duplicates.

3. FROM
   - Specifies the table from which data is retrieved.

4. WHERE
   - Filters records based on specified conditions.

5. AND
   - Returns records when all specified conditions are TRUE.

6. OR
   - Returns records when at least one condition is TRUE.

7. NOT
   - Negates a condition.

8. IN
   - Checks whether a value matches any value in a given list.

9. NOT IN
   - Excludes records matching values in a given list.

10. BETWEEN
    - Filters values within a specified range (inclusive).

11. LIKE
    - Searches for a specified pattern in a column.

12. IS NULL
    - Checks whether a column contains NULL values.

13. IS NOT NULL
    - Checks whether a column does not contain NULL values.

14. ORDER BY
    - Sorts records in ascending (ASC) or descending (DESC) order.

15. ASC
    - Sorts records in ascending order (default).

16. DESC
    - Sorts records in descending order.

17. LIMIT
    - Restricts the number of rows returned by a query.

18. OFFSET
    - Skips a specified number of rows before returning results.

19. GROUP BY
    - Groups rows having the same values in specified columns.

20. HAVING
    - Filters grouped records based on aggregate conditions.

21. AS
    - Assigns a temporary alias to a column or table.

22. CASE
    - Performs conditional logic in SQL (similar to if-else).

23. WHEN
    - Specifies a condition inside a CASE expression.

24. THEN
    - Specifies the result when a CASE condition is TRUE.

25. ELSE
    - Specifies the alternative result when no CASE condition matches.

26. END
    - Ends a CASE expression.

27. WITH
    - Defines a Common Table Expression (CTE) for a query.

28. UNION
    - Combines results of multiple SELECT queries and removes duplicates.

29. UNION ALL
    - Combines results of multiple SELECT queries without removing duplicates.

30. INTERSECT
    - Returns common rows from multiple SELECT queries.

31. EXCEPT
    - Returns rows from the first query that are not present in the second.

========================================================
                    JOINS
========================================================

32. INNER JOIN
    - Returns matching records from both tables.

33. LEFT JOIN
    - Returns all records from the left table and matching records
      from the right table. Non-matching right-side values are NULL.

34. RIGHT JOIN
    - Returns all records from the right table and matching records
      from the left table. Non-matching left-side values are NULL.

35. CROSS JOIN
    - Returns the Cartesian product of two tables.

36. SELF JOIN
    - Joins a table with itself.

37. ON
    - Specifies the condition used to join two tables.

38. USING
    - Specifies common column names used to join two tables.

========================================================
                AGGREGATE FUNCTIONS
========================================================

39. COUNT()
    - Returns the number of rows or non-NULL values.

40. SUM()
    - Returns the total sum of numeric values.

41. AVG()
    - Returns the average of numeric values.

42. MIN()
    - Returns the minimum value.

43. MAX()
    - Returns the maximum value.

44. COUNT(DISTINCT)
    - Counts unique non-NULL values.

========================================================
                WINDOW FUNCTIONS
========================================================

45. OVER()
    - Defines the window of rows used by a window function.

46. PARTITION BY
    - Divides rows into groups for window function calculations.

47. ROW_NUMBER()
    - Assigns a unique sequential number to each row.

48. RANK()
    - Assigns ranks with gaps when duplicate values occur.

49. DENSE_RANK()
    - Assigns ranks without gaps when duplicate values occur.

50. LAG()
    - Accesses a value from a previous row.

51. LEAD()
    - Accesses a value from a following row.

52. FIRST_VALUE()
    - Returns the first value in a window.

53. LAST_VALUE()
    - Returns the last value in a window.

========================================================
              DATA DEFINITION LANGUAGE (DDL)
========================================================

54. CREATE
    - Creates databases, tables, views, and other objects.

55. ALTER
    - Modifies the structure of an existing database object.

56. DROP
    - Permanently deletes a database object.

57. TRUNCATE
    - Removes all rows from a table while keeping its structure.

58. RENAME
    - Renames an existing database object or table.

========================================================
              DATA MANIPULATION LANGUAGE (DML)
========================================================

59. INSERT INTO
    - Inserts new records into a table.

60. UPDATE
    - Modifies existing records in a table.

61. DELETE
    - Deletes specified records from a table.

========================================================
              DATA CONTROL LANGUAGE (DCL)
========================================================

62. GRANT
    - Gives specified privileges to database users.

63. REVOKE
    - Removes specified privileges from database users.

========================================================
              TRANSACTION CONTROL LANGUAGE (TCL)
========================================================

64. COMMIT
    - Permanently saves the current transaction changes.

65. ROLLBACK
    - Undoes uncommitted transaction changes.

66. SAVEPOINT
    - Creates a point within a transaction to which you can roll back.

67. START TRANSACTION
    - Begins a new transaction.

========================================================
              IMPORTANT CONSTRAINTS
========================================================

68. PRIMARY KEY
    - Uniquely identifies each row in a table.
    - Cannot contain NULL values.

69. FOREIGN KEY
    - Creates a relationship between tables by referencing
      a key in another table (or the same table).

70. UNIQUE
    - Ensures that values in a column or combination of columns
      are unique.

71. NOT NULL
    - Prevents a column from storing NULL values.

72. CHECK
    - Ensures that values satisfy a specified condition.

73. DEFAULT
    - Assigns a default value when no value is provided.

74. AUTO_INCREMENT
    - Automatically generates sequential numeric values
      for a column in MySQL.

========================================================
              SUBQUERY OPERATORS
========================================================

75. EXISTS
    - Checks whether a subquery returns at least one row.

76. NOT EXISTS
    - Checks whether a subquery returns no rows.

77. ANY
    - Compares a value with any value returned by a subquery.

78. ALL
    - Compares a value with all values returned by a subquery.

79. SOME
    - Works like ANY in SQL comparisons.

========================================================
              OTHER IMPORTANT KEYWORDS
========================================================

80. DESC (DESCRIBE)
    - Displays the structure of a table in MySQL.

81. SHOW
    - Displays database information such as databases and tables.

82. USE
    - Selects a database to work with.

83. INDEX
    - Helps speed up data retrieval from a table.

84. VIEW
    - Creates a virtual table based on a SELECT query.

85. IF EXISTS
    - Checks
