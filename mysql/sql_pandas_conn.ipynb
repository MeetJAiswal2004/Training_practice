#### database metal with table = diamond_data

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("metal.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS diamonds_data (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    carat REAL,
    cut TEXT,
    color TEXT,
    clarity TEXT,
    price REAL
)
""")
```

```python
df = pd.read_sql_query("SELECT * FROM diamonds_data", conn)
display(df)
```

```python
cursor.executemany("""
INSERT INTO diamonds_data (carat, cut, color, clarity, price)
VALUES (?, ?, ?, ?, ?)
""", [
    (0.23, 'Ideal', 'E', 'SI2', 326.00),
    (0.21, 'Premium', 'E', 'SI1', 326.00),
    (0.29, 'Premium', 'I', 'VS2', 334.00),
    (0.31, 'Good', 'J', 'SI2', 335.00)
])
conn.commit()
```

```python
df = pd.read_sql_query("SELECT * FROM diamonds_data", conn)
display(df)
```

```python
## delete each data inside the table, but keep the table 
cursor.execute("DELETE FROM diamonds_data")
conn.commit()
```

```python
## drop the entire table 
cursor.execute("DROP TABLE diamonds_data")
conn.commit()
```

```python
### close the connection first before dropping the entire database 
conn.close()

### then remove that databse from the os memory also 
import os
if os.path.exists("diamond.db"):
    os.remove("diamond.db")
    print("Database diamond.db deleted successfully!")
```

#### database diamond with table employee

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("diamond.db")

cursor = conn.cursor()

cursor.execute("""
    CREATE TABLE IF NOT EXISTS employee (
        emp_id INT PRIMARY KEY,
        name VARCHAR(50),
        salary FLOAT(30),
        exp FLOAT(10)
    )
""")
```

```python
df = pd.read_sql_query("SELECT * FROM employee", conn)
display(df)
```

```python
cursor.executemany("""
    INSERT INTO employee (emp_id, name, salary, exp) VALUES (?,?,?,?)
""", [
        (1,'meet jaiswal',400000.00,2),
        (2,'dev jaiswal',50000.00,3),
        (3,'raj solanki',20000.00,4)
])
conn.commit()
```

```python
# 1. Always close the connection first (to release the lock)
conn.close()

# 2. Delete the database file using Python
import os

if os.path.exists("metal.db"):
    os.remove("metal.db")
    print("Database diamond.db deleted successfully!")
```
