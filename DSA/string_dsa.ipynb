- reverse string

```python
## slicing, ## Time Complexity: O(n), Space Complexity: O(n)
s = "meetjasiwal"

def reverse_string(s):
    return s[::-1]

print(reverse_string(s))
```

```python
## two- pointer 
## Time Complexity: O(n), Space Complexity: O(n)
s = "jasiwal"
def reverse_str(s):
    lst = list(s)
    i = 0
    j = len(lst)-1

    while i<j:
        lst[i],lst[j] = lst[j],lst[i]
        i += 1
        j -= 1

    return "".join(lst) ## join() by default return string

print(reverse_str(s))

```

```python
## using reversed()
s = "string"
def reverse_str(s):
    # return "".join(reversed(s))
    hello = "".join(reversed(s))
    return hello

print(reverse_str(s))
```

```python
## using backward loop
def reverse_str4(s):
    res = []
    for i in range(len(s) - 1, -1, -1):
        res.append(s[i])
    return "".join(res)
print(reverse_str4(s))
```

- check palindrome

```python
## slicing method #Time: 𝑂(𝑛) | Space: 𝑂(n)
s = "aabbaa"

def is_palindrome(s):
    if s == s[::-1]:
        return True
    return False

print(is_palindrome(s))
```

```python
# two popinter method
#Time: 𝑂(𝑛) | Space: 𝑂(1) (No extra memory)
s = "aaabbccbbaaa"
def is_palindrome(s):
    lst = str(s)
    left = 0
    right = len(s)-1
    while left<right:
        if lst[left] != lst[right]:
            return False
        left += 1
        right -= 1
    return True

print(is_palindrome(s))
```

```python
## half loop Time: t:O(n), s:O(1)

s = "mmeemm"

def is_palindrome(s):
    lst = list(s)
    for i in range(len(lst)//2):
        if lst[i] != lst[len(lst)-1-i]:
            return False
    return True

print(is_palindrome(s))
```
