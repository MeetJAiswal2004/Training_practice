- find the total of the each element

```python
nums = [10,20,30,40,50]

## built in python function (sum()), t.c = o(n), s.c=o(1) (efficient)
def total_sum1(nums):
    return sum(nums)
print(total_sum1(nums))

## using for loop (basic approach), t.c = o(n), s.c = o(1) (efficient)
def total_sum2(nums):
    total = 0 
    for num in nums:
        total += num
    return total
print(total_sum2(nums))

## using recursion ,Time Complexity: O(n²) due to list slicing. Space Complexity: O(n) due to recursion and slices.
def total_sum3(nums):
    if len(nums)==0:
        return 0

    return nums[0]+ total_sum3(nums[1:])  
print(total_sum3(nums))
```

- Find the largest element in a list

```python
nums = [10,22,39,5,55,68,40,30]

## using python built in function max()
def find_max(nums):
    result = max(nums)
    return result
print(find_max(nums))

## using for loop
def find_max2(nums):
    max = nums[0]
    for num in nums:
        if max < num: ## both conditions true 1) max<num or 2)num>max
            max = num
    return max
print(find_max2(nums))

## sort list and print the last element
def find_max3(nums):
    nums.sort()
    return nums[-1]
print(find_max3(nums))
```

- find the smallest element in the list 

```python
nums = [10,22,39,5,55,68,40,30]

def find_min(nums):
    result = min(nums)
    return result
print(find_min(nums))

def find_min2(nums):
    min = nums[0]
    for num in nums:
        if num<min:
            min = num
    return min
print(find_min2(nums))

def find_min3(nums):
    nums.sort()
    return nums[0]
print(find_min3(nums))
```

- count even and odd

```python
lst = [10,3,5,6,77,89,88,2]

## using for loop
def even_odd(lst):
    even = 0
    odd = 0
    for num in lst:
        if num % 2 == 0:
            even += 1
        else:
            odd += 1
    return (even,odd)
print(even_odd(lst))

## using while loop
def even_odd2(lst):
    even=0 
    odd = 0
    i = 0
    while i < len(lst):
        if lst[i] % 2 == 0:
            even += 1
        else:
            odd += 1
        i += 1
    return (even,odd)
print(even_odd2(lst))

## remove odd numbers inplace in the list (only even)
def only_even(lst):
    i = 0
    while i < len(lst):
        if lst[i] % 2 != 0:
            lst.pop(i)
        else:
            i += 1
    return lst
print(only_even(lst))

## store even elements in the new list
def only_even2(lst):
    even = []
    for num in lst:
        if num % 2 == 0:
            even.append(num)
    return even
print(only_even2(lst)) 

## remove even elements (only odd ) in place
def odd_only(lst):
    i = 0
    while i < len(lst):
        if lst[i] % 2 == 0:
            lst.pop(i)
        else:
            i += 1
    return lst
print(odd_only(lst))
```

- flatten deep nested list /nested list

```python
# flatten deep nested list
lst = [1,[2, [3, [4, [5,[6,7]]]]]]


flat =[]

# solution 1 (less complex using isinstance())

def flatten(lst,flat):
    for n in lst:
        if isinstance(n,list):
            flatten(n,lst)
        else:
            flat.append(n)
    return flat

print(flatten(lst,flat))

```

```python
# solution 2 (having more complex than 1st solution ,using type())
lst = [1,[2, [3, [4, [5,[6,7]]]]]]

flat =[]

def flatten1(lst,flat):
    for n in lst:
        if type(n) == list:
            flatten1(n,lst)
        else:
            flat.append(n)
    return flat

print(flatten1(lst,flat))
```

- converting flatten list into deep nested list

```python
lst = [1,2,3,4,5,6,7]
# target = [1,2[3,4,[5,6,[7]]]] 
def task(lst):
    nestlist=[lst[0],lst[1]]
    n = len(lst)
    current = nestlist
    for i in range(2,n,2):
        temp = [lst[i]]

        if i + 1 < n:
            temp.append(lst[i+1])

        current.append(temp)
        current = temp
    return nestlist

print(task(lst))
```

- frequency check 

```python
lst = [1,2,4,1,3,2,4,5,5,6,1,4]
freq ={}
def frequency(lst,freq):
    for x in lst:
        if x not in freq:
            freq[x] = 1
        else:
            freq[x] += 1
    return freq

print(frequency(lst,freq))
```

```python
## frequnecy check using builtin module
from collections import Counter

lst = [1,2,4,1,3,2,4,5,5,6,1,4]

def  frequency(lst):
    result = Counter(lst)
    return result

print(frequency(lst))
```

- common elements


```python
l1 = [1,2,4,8,5,2,5,7]
l2 = [6,3,8,1,6,3,9,5,3]
common = []

def common_element(l1,l2,common):
    for n in l1:
        if n in l2 and n not in common:
            common.append(n)
    return common

print(common_element(l1,l2,common)) 
```
