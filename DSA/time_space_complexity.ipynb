##  Time complexity efficiency

| Big O Notation | Name | Description | Example / Typical Context |
|---|---|---|---|
| **O(1)** | Constant Time | Runtime stays the same regardless of input size | Accessing an array element by index, hash table lookup |
| **O(log n)** | Logarithmic Time | Runtime increases very slowly as input grows | Binary search, balanced search tree operations |
| **O(n)** | Linear Time | Runtime grows directly with input size | Single loop iterating through a list, finding max/min element |
| **O(n log n)** | Linearithmic Time | Growth is slightly worse than linear; typical of efficient divide-and-conquer algorithms | Merge sort, quicksort (average case), heapsort |
| **O(n²)** | Quadratic Time | Runtime scales with the square of the input size (often nested loops) | Bubble sort, selection sort, insertion sort (average/worst case) |
| **O(n³)** | Cubic Time | Runtime scales with the cube of the input size (triple-nested loops) | Naive matrix multiplication |
| **O(2ⁿ)** | Exponential Time | Runtime doubles with each additional input element; becomes impractical quickly | Recursive Fibonacci without memoization, subset generation |
| **O(n!)** | Factorial Time | Extremely slow; usable only for very small datasets | Traveling Salesperson Problem (brute-force permutation search) |


## Efficiency Hierarchy Summary

O(1) < O(log n) < O(n) < O(n log n) < O(n²) < O(n³) < O(2ⁿ) < O(n!)
